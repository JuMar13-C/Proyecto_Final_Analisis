El original completo no pudo descargarse. Este notebook conserva las instrucciones de recuperación mostradas en la captura y añade un bucle, rutas y validación. En Colab sube Paquete_Encoded_FFT.zip; descarga Paquete_Recuperados.zip.

```python
import numpy as np
from scipy.io.wavfile import read, write
from scipy.fft import rfft, irfft
from IPython.display import Audio, display
from pathlib import Path
import zipfile
BASE=Path.cwd()
if not (BASE/'Codificados/encoded1.wav').exists():
    (BASE/'Codificados').mkdir(exist_ok=True)
    with zipfile.ZipFile(BASE/'Paquete_Encoded_FFT.zip') as z:
        assert z.testzip() is None
        for i in range(1,6):
            for name in (f'encoded{i}.wav',f'FFT_key{i}.csv'):
                (BASE/'Codificados'/name).write_bytes(z.read(name))
(BASE/'Recuperados').mkdir(exist_ok=True)
for i in range(1,6):
    # Núcleo Reverse de la captura docente, adaptado solo para cinco rutas.
    fs, encoded = read(BASE/'Codificados'/f'encoded{i}.wav')
    key = np.loadtxt(BASE/'Codificados'/f'FFT_key{i}.csv', delimiter=',', skiprows=1, dtype=int)
    X_encoded = rfft(encoded.astype(float))
    assert np.array_equal(np.sort(key), np.arange(len(X_encoded)))
    X_recovered = np.empty_like(X_encoded)
    X_recovered[key] = X_encoded
    recovered = irfft(X_recovered, n=len(encoded))
    write(BASE/'Recuperados'/f'recovered{i}.wav', fs, recovered.astype(np.float32))
    print(f'Recuperado {i}: {len(recovered)} muestras')
    display(Audio(recovered, rate=fs))
with zipfile.ZipFile(BASE/'Paquete_Recuperados.zip','w',zipfile.ZIP_DEFLATED) as z:
    for i in range(1,6):
        name=f'recovered{i}.wav'; z.write(BASE/'Recuperados'/name,name)
```
